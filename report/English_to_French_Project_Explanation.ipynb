# English-to-French Deep Learning Translation Project

“My project is an English-to-French machine translation application built with Python, NLP, Hugging Face Transformers, and Streamlit. Initially, I developed an encoder–decoder LSTM model. The encoder processed English token sequences and represented them using hidden and cell states; the decoder learned to predict the next French token. However, the generated translations were not reliable enough, so I changed the application’s translation engine to the pretrained model `Helsinki-NLP/opus-mt-en-fr` from the Hugging Face Hub.

The current model is based on the Marian encoder–decoder Transformer architecture. In `translator.py`, I load the Marian tokenizer and model using `from_pretrained()`. The tokenizer converts the English sentence into model-readable token IDs. The model generates French token IDs, and the tokenizer decodes them into a readable French sentence. The Streamlit application in `app.py` calls `translate_sentence()` and displays the result.

The main advantage is that I can use translation knowledge already learned by a pretrained model instead of training all model parameters from scratch. The current version performs inference; I have not fine-tuned Marian on my local `fra.txt` dataset. I would evaluate it using a representative test set, reference-based metrics such as BLEU or chrF, inference time, model size, and human review. I would describe cloud deployment as complete only after testing the public Streamlit Community Cloud URL.”

### How the system works

**English input → tokenizer → pretrained Marian encoder–decoder Transformer → generated token IDs → tokenizer decode → French output → Streamlit display**

1. **Streamlit (`app.py`)** provides the user interface.
2. **`translator.py`** contains the translation function and loads the model/tokenizer.
3. **Tokenizer** converts text into model-readable tokens and IDs.
4. **Marian Transformer** generates a French sequence using pretrained weights.
5. **Tokenizer decoding** turns output IDs into readable French text.

### What is implemented vs. future work

**Implemented / reported working**
- Local Streamlit translation interface.
- `translator.py` uses `MarianTokenizer` and `MarianMTModel`.
- Pretrained `Helsinki-NLP/opus-mt-en-fr` is used for inference.
- Local translations are reported to be mostly correct.

**Not yet confirmed or not implemented in the described workflow**
- Fine-tuning Marian on `fra.txt`.
- Formal BLEU/chrF benchmark or controlled LSTM-versus-Marian comparison.
- Verified Streamlit Community Cloud deployment.
- Sentence-embedding/cosine-similarity feature integrated into the app.
- A separate BERT experiment.

### Why Hugging Face?

“I used Hugging Face because it provides a model hub and a Python library for loading pretrained models. This let me use a dedicated translation model without training all its parameters from scratch. I can identify the exact model by its repository name, inspect its model card, load it through the Transformers API, and evaluate it on my own test sentences.”

### Is it fine-tuned?

“No. The current version loads the pretrained Marian model and uses it for inference. Fine-tuning would be a separate step using suitable aligned English–French examples, followed by evaluation on held-out examples.”

### How will you prove it works?

“I will test varied English sentences against human-checked French references, review meaning and grammar, calculate BLEU or chrF where references are available, measure inference latency separately from model startup, and record model size. I will not claim a numerical accuracy until I have measured it.”
